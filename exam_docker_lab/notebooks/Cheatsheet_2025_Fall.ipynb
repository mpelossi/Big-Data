### Orders Dataset

We will use a dataset that contains an array of orders with respective customer, order date, purchased items, etc. Each order has all of the following attributes:

- `order_id`: an integer that uniquely identifies an order
- `date`: string containing the order date
- `customer`:
    - `first_name`: string containing the first name of a customer
    - `last_name`: string containing the last name of a customer
- `items`: array of ordered items, each of which is a dictionary with the following key-value pairs:
    - `product`: string with the product name
    - `price`: floating point number with the price of the item
    - `quantity`: an integer with the number of times this product is ordered

### DataFrame interface

Run the code below to create a Spark DataFrame:

```python
%load_ext sparksql_magic

orders_df = spark.read.json('orders.jsonl')
orders_df.createOrReplaceTempView("orders")
```

Print the type and schema:


```python
orders_df.printSchema()
```

Print one row using spark dataframe.

```python
%%sparksql
SELECT customer, sum(items_new.quantity) as sum_num
FROM (SELECT *
FROM (
    SELECT order_id, date, customer, explode(items) as items_new
    FROM orders
))
WHERE items_new.product ='computer screen'
GROUP BY customer
ORDER BY sum(items_new.quantity) desc
LIMIT 1
```

```python
%%sparksql 

SELECT first(customer), SUM(flat_items.quantity) AS q 

FROM orders lateral VIEW EXPLODE(items) AS flat_items 

WHERE flat_items.product == "computer screen"

GROUP BY order_id

ORDER BY q DESC
```

## General Queries

To print the dataset, use *json-file()*.

```python
%%jsoniq
json-file("git-archive.json", 10)[1]
```

To see the fields, use the *keys()* function.

```python
%%jsoniq
for $e in json-file("git-archive.json", 10)
order by size($e.payload.commits) descending
count $c
where $c < 2
return ($e.payload.commits[].author.name)
```

You can display the first objects with the *count* clause.

```python
%%jsoniq
for $e in json-file("git-archive.json", 10)
where $e.type eq "PushEvent"
order by size($e.payload.commits) descending
count $c
where $c < 2
return ($e.payload.commits[].author.name)
```

You can use . for object navigation,  [] for navigation (in parallel).

```python
%%jsoniq
for $e in json-file("git-archive.json", 10)
where $e.type eq "PushEvent"
for $c in $e.payload.commits[]
group by $d := $c.author.name 
order by count($c) descending
count $g
where $g eq 1
return $d
```

```python
%%jsoniq
for $e in json-file("git-archive.json", 10)
for $c in $e.payload.commits[]
group by $d := $c.author.name 
order by count($c) descending
count $g
where $g eq 1
return $d
```

A more complex query with FLWOR:

When using `%%jsoniq -pdf`, the result is printed as a dataframe. Without `-pdf`, it is printed in JSON.

```python
%%jsoniq -pdf

for $e in json-file("git-archive.json", 10)
let $actor := $e.actor.id
group by $actor
let $nb_commits := count($e.payload.commits[])
order by $nb_commits descending
count $n
where $n le 15
return {
    "actor" : $actor,
    "number of commits" : $nb_commits,
    "first commit email" : $e[1].payload.commits[[1]].author.email
}
```
